```python
# Analizar tipos de datos
data3.dtypes
```

```python
data3["Potencial de compra"] = pd.to_numeric(data3["Potencial de compra"], errors='coerce')
```

```python
# Remplazar con la mediana
# COLUMNA POTENCIAL DE COMPRA
data3["Potencial de compra"]=data3["Potencial de compra"].fillna(round(data3["Potencial de compra"].median(),1))
```

```python
# Remplazar con media
# COLUMNAS PDM, SDC y VENTA
data3["Venta"]=data3["Venta"].fillna(round(data3["Venta"].mean(),1))
```

```python
data3 = data3.rename(columns={'Unnamed: 15': 'Comentarios'})
```

```python
#Sustituir valores nulos por un string en concreto 
data3["BDC "] =data3["BDC "].fillna("Sin BDC asignado")
data3["Fecha"] =data3["Fecha"].fillna("2030-06-19 00:00:00")
data3["Nombre Cliente "] =data3["Nombre Cliente "].fillna("Sin nombre especificado")
data3["Telefono "] =data3["Telefono "].fillna("Sin telefono")
data3["Asesor Asignado"] =data3["Asesor Asignado"].fillna("Sin asesor asignado")
data3["¿Por que no ha visitado la agencia?"] =data3["¿Por que no ha visitado la agencia?"].fillna("Sin razón especificada") 
data3["¿Por que no hay solicitud de credito?"] =data3["¿Por que no hay solicitud de credito?"].fillna("Sin razón especificada") 
data3["¿Por que no hubo prueba de manejo ?"] =data3["¿Por que no hubo prueba de manejo ?"].fillna("Sin razón especificada")
data3["¿Por que no hubo proceso wow?"] =data3["¿Por que no hubo proceso wow?"].fillna("Sin razón especificada") 
data3["¿Por que se asigno antes de vistar piso?"] =data3["¿Por que se asigno antes de vistar piso?"].fillna("Sin razón especificada")
data3["Comentarios"] =data3["Comentarios"].fillna("Sin comentarios")
```

```python
# Verificar la columna "Estatus de Lead" para ver los valores únicos
data3["Estatus de Lead"].unique()
```

```python
# Homologar los valores de la columna "Estatus de Lead" para que sean consistentes
data3['Estatus de Lead'] = data3['Estatus de Lead'].astype(str).str.strip()

homologados = {
    # Variantes de Activo
    'Activo': 'Activo', 'Acitvo': 'Activo', 'Acrivo': 'Activo', 
    'activo': 'Activo', 'Ativo': 'Activo', 'Activio': 'Activo', 'Activo-': 'Activo',
    # Variantes de Venta
    'Venta': 'Venta',
    # Variantes de Finalizado
    'Finalizado': 'Finalizado'
}
data3['Estatus de Lead'] = data3['Estatus de Lead'].replace(homologados)
```

```python
data3["Estatus de Lead"].unique()
```

```python
# Remplazar valores nulos por un string en concreto 
data3["Estatus de Lead"] =data3["Estatus de Lead"].fillna("Desconocido") 
```
