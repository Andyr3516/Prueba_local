```python
data2=data.copy()
```

```python
data2.dtypes
```

```python
data2['Telefono'] = data2['Telefono'].astype(str).str.strip()
data2['Correo'] = data2['Correo'].astype(str).str.strip()
```

```python
data2["Telefono"] = data2["Telefono"].fillna("Sin especificar") 
data2["Correo"] = data2["Correo"].fillna("Sin especificar")
data2["Producto"] = data2["Producto"].fillna("Sin especificar") 
```

```python
#Corroborar valores nulos por dataframe
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Convertir DataFrame a CSV
data2.to_csv("base_leads_sin_nulos.csv", index=False)
```
